```python
#ejm 6
def  dollars_to_euro(dollar_amount):
  print(dollar_amount * 0.93)
```

```python
# we need to add the argument to value at dollar_amount
dollars_to_euro(5.0)
dollars_to_euro(10)
dollars_to_euro(20.0)
```

```python
#ejm 7 multiple parameters
def dollars_to_euro_with_rate(dollar_amount,convertion_rate):
  print(dollar_amount * convertion_rate)
```

```python
dollars_to_euro_with_rate(5.0,0.93)
dollars_to_euro_with_rate(10,0.93)
dollars_to_euro_with_rate(20.0,0.93)
#dollars_to_euro_with_rate(20.5) this will send error, we need both values
```

```python
#ejm 8 default parameters Values
def dollars_to_euro_with_default(dollar_amount,convertion_rate=0.93):
  print(dollar_amount * convertion_rate)
```

```python
dollars_to_euro_with_default(10.0)
dollars_to_euro_with_default(10.0,0.88)
```

```python
#ejm 9 expression return (Si queremos que la función devuelva uno o varios valores, podemos usar return)
def dollars_to_euro_with_return(dollar_amount,convertion_rate=0.93):
  return dollar_amount * convertion_rate
```

```python
a = dollars_to_euro_with_default(19.0)
print(a)
```

```python
#function with docstring
def dollars_to_euro_with_doc(dollar_amount: float= 5.5 ,convertion_rate: float = 0.93):
    """
    Return dollar amount to euros based on a conversion rate.

    Parameters:
      Dollar_amount(float): Dollar amount to be converted to euros
      conversion_rate (float): Dollar to euro conversion rate. Default:0.93

    Returns:
      euro_amount (float): Euro equivalent of dollar amount based on convertion rate

     """
    euro_amount = dollar_amount * convertion_rate
    return euro_amount
    print(euro_amount)
```

```python
euro_amount = dollars_to_euro_with_doc(5.0)
print(euro_amount)
```
